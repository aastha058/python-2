## Abstraction in Python


#### Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object.This help in reducing programming complexity and effort.

```python
from abc import ABC, abstractmethod

# Abstraction base class
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving")

    @abstractmethod
    def start_engine(self):
        pass

# Derived class
class Car(Vehicle):
    def start_engine(self):
        print("The car engine has started")

# Function that operates any vehicle
def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

# Create an object and operate it
car = Car()
operate_vehicle(car)
```
