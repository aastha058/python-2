```python
## Instance variable and method

class dog:
    ## constructor
    def __init__(self,name,age,roll):
        self.name=name
        self.age=age
        self.roll=roll

## create a object 
dog1=dog("buddy",3,4)
print(dog1)
print(dog1.name)
print(dog1.age)
print(dog1.roll)
```

```python
## one more create a object

dog2=dog("lucy",4,23)
print(dog2)
print(dog2.name)
print(dog2.age)
print(dog2.roll)
```

```python
## define a class with instance method

class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof!")



dog1 = Dog("Buddy", 3)
dog1.bark()
dog2 = Dog("luck",4)
dog2.bark()
```

```python
## modeling a bank account

# difing a class for bank account

class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} is deposited. New balance is {self.balance}")
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")
    def get_balance(self):
        return self.balance

## create a object and an account

account = BankAccount("Aastha", 20000)
print(account.balance)
```

```python
## call isntance method 
account.deposit(100)
```

```python
account.withdraw(300)
```

```python
print(account.get_balance())
```
