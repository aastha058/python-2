```python
## some basic magic methods are:

class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def __str__(self):
        return f"Person name is {self.name} and age is {self.age}"
    def __repr__(self):
        return f"Person(name={self.name},age={self.age})"


person=Person("aastha",21)  
print(person)  # it will call __str__ method
print(repr(person))  # it will call __repr__ method
```
