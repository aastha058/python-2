### polymorphism

```python
# base class
class Animal:
    def speak(self):
        return "sound of the animal"

# derived class 1
class Dog(Animal):
    def speak(self):
        return "woof!"

# derived class 2
class Cat(Animal):
    def speak(self):
        return "meow!"

# function that demonstrates polymorphism
def animal_speak(animal):
    print(animal.speak())

# create objects
dog = Dog()
cat = Cat()

# direct method calls
print(dog.speak())
print(cat.speak())

# using the polymorphic function
animal_speak(dog)
animal_speak(cat)
```
