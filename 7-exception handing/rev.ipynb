```python

# 📝 Simple To-Do List Program using Loops

tasks = []  # खाली लिस्ट, इसमें सारे काम (tasks) रखेंगे

while True:
    print("\n====== TO-DO LIST MENU ======")
    print("1. नया Task जोड़ें")
    print("2. सारे Tasks देखें")
    print("3. कोई Task हटाएँ")
    print("4. प्रोग्राम बंद करें")

    choice = input("अपना चुनाव दर्ज करें (1-4): ")

    # ✅ 1️⃣ नया Task जोड़ना
    if choice == "1":
        task = input("अपना नया Task लिखें: ")
        tasks.append(task)
        print(f"✅ '{task}' लिस्ट में जोड़ दिया गया!")

    # ✅ 2️⃣ सारे Task दिखाना
    elif choice == "2":
        print("\n📋 आपकी To-Do List:")
        if len(tasks) == 0:
            print("(अभी कोई Task नहीं है)")
        else:
            for i, t in enumerate(tasks, start=1):
                print(f"{i}. {t}")

    # ✅ 3️⃣ Task हटाना
    elif choice == "3":
        if len(tasks) == 0:
            print("कोई Task मौजूद नहीं है!")
        else:
            for i, t in enumerate(tasks, start=1):
                print(f"{i}. {t}")
            num = int(input("कौन-सा Task हटाना है (नंबर डालें): "))
            if 1 <= num <= len(tasks):
                removed = tasks.pop(num - 1)
                print(f"🗑 '{removed}' हटा दिया गया!")
            else:
                print("❌ गलत नंबर डाला गया!")

    # ✅ 4️⃣ प्रोग्राम बंद करना
    elif choice == "4":
        print("👋 धन्यवाद! आपका दिन शुभ हो!")
        break

    # ❌ गलत इनपुट पर संदेश
    else:
        print("⚠️ कृपया 1 से 4 के बीच का विकल्प चुनें!")
```
